```python
import numpy as np 
```

```python
data = np.genfromtxt(
    "data.csv",
    delimiter = ',',
    names = True,
    dtype = None,
    encoding = 'utf-8'
)
print(data)
```

```python
print(data.dtype.names)
```

```python
print(data["Age"])
```

```python
print(data["Product"])
```

```python
ages = data["Age"]
quantity = data["Quantity"]
prices = data["Unit_Price"]
rating = data["Rating"]
genders = data["Gender"]
products = data["Product"]
```

```python
print(ages)
```

```python
print(quantity)
```

```python
print(prices)
```

```python
print(rating)
```

```python
# calculate sales
sales = quantity * prices
```

```python
print(sales)
```

```python
print("/n---Age----")
print(f"Average age: {np.average(ages)}")
print(f"meadian age: {np.median(ages)}")
print(f"youngest age: {np.min(ages)}")
print(f"oldest age: {np.max(ages)}")
```

```python
print("\n--- Sales ---")
print("Total sales:", np.sum(sales))
print("Average sale:", np.mean(sales))
print("Median sale:", np.median(sales))
print("Minimum sale:", np.min(sales))
print("Maximum sale:", np.max(sales))
print("Sales standard deviation:", np.std(sales))
```

```python
print("\n--- Quantity ---")
print("Total items sold:", np.sum(quantity))
print("Average quantity:", np.mean(quantity))
```

```python
print("\n--- Ratings ---")
print("Average rating:", np.mean(rating))
print("Highest rating:", np.max(rating))
print("Lowest rating:", np.min(rating))
```

```python
print("\n--- Gender ---")
print("Male customers:", np.sum(genders == "Male"))
print("Female customers:", np.sum(genders == "Female"))
```

```python
print("\n--- Age Analysis ---")
print("Customers over 40:", np.sum(ages > 40))
```

```python
print("\n--- Sales Analysis ---")
print("Orders above RS500:", np.sum(sales > 500))
```

```python
# Highest sale
index = np.argmax(sales)
print(index)
```

```python

print("\nHighest sale:")
print("Order:", data["Order_ID"][index])
print("Product:", products[index])
print("Amount:", sales[index])
```
