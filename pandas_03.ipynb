```python
# merge and join in pandas
df_customer = pd.DataFrame({
    "customer_id":[1,2,3,4],
    "customer_name": ["irfan","humair","hamza","maaz"]
})

df_order = pd.DataFrame({
    "order_id":[101,102,103,104],
    "customer_id":[1,4,5,3],
    "amount": [250,450,340,230]
})
```

```python
df_customer
```

```python
df_order
```

```python
pd.merge(df_customer, df_order, on="customer_id")
```

```python
pd.merge(df_customer, df_order, on="customer_id", how="right")
```

```python
pd.merge(df_customer, df_order, on="customer_id", how="left")
```

```python
pd.merge(df_customer, df_order, on="customer_id",how="outer")
```

```python
pd.concat([df_customer,df_order],ignore_index=True)
```

```python
pd.concat([df_customer, df_order],axis=1)
```
