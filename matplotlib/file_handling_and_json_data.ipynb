## File Handling

```python
file = open('student.txt', 'r')
print(file.read())
print(f"Mode of the File:{file.mode}")
print(f"the file is closed:{file.closed}")
file.close()
print(f"the file is closed:{file.closed}")
print(f"file name: {file.name}")
```

```python
with open('student.txt', 'r') as file:
    content = file.read()
    print(content)
```

```python

with open('student.txt','r') as file:
    print(file.readlines())
```

```python
with open('student.txt', 'r') as file:
    file.seek(0)
    content = file.read()
print(content)
```

```python
with open("student.txt","w") as file:
    file.write("hellow world")
```

```python
with open("student.txt", "a") as file:
    file.tell()
    file.write("\nahmad")
```

```python
with open('student.txt', 'w+') as file:
    file.write("wellcome to Aptech Metro StarGate\n")
    file.write("my name is ahmad khan")
    content = file.readlines()
    
print(content)    
```

```python
with open("student.txt",'r+') as file:
    content = file.read()
print(content)
```

## Json In Python

```python
import json
data = {
    "Name":"Bilal",
    "age":21,
    "city":"Karachi"
}
print(data, type(data))
json_data = json.dumps(data)
print(json_data, type(json_data))
```

```python
json_data = '{"Name":"ali","age":23}'
print(json_data, type(json_data))

std_data = json.loads(json_data)
print(std_data, type(std_data))
```

```python
data = [
    {
        "Name":"Bilal",
        "Age":23,
        "City":"Karachi",
        "Mark":87
    },
    {
        "Name":"Ahmad",
        "Age":23,
        "City":"Lahore",
        "Mark":85
    },
    {
        "Name":"Waqar",
        "Age":24,
        "City":"Swat",
        "Mark":90
    },
    {
        "Name":"Shabir",
        "Age":23,
        "City":"Karachi",
        "Mark":87
    },
]
with open('student.json', 'w') as file:
    json.dump(data, file, indent=4)
```

```python
with open("student.json",'r') as file:
    content = json.load(file)

for data in content:
    print(data['Age'] >= 24)
```
