```python
import matplotlib.pyplot as plt
```

```python
x = [10,20,30,40]
y = [15,25,35,45]
plt.plot(x,y)
plt.show()
```

```python
plt.figure(figsize=(4,5))
plt.plot(x,y, color="blue", marker="o", linestyle="--", linewidth=2, markersize = 12)
plt.title("figure size title")
plt.xlabel("represent x label")
plt.ylabel("y label represent")
plt.show()
```

```python
years = [2021,2022, 2023, 2024, 2025]
y1 = [10,30,40,50,60]
y2 = [20,35,45,55,65]
plt.plot(years,y1, label="2021-sales")
plt.plot(years,y2, label="2022-sales")
plt.xlabel("Year")
plt.ylabel("sales")
plt.legend()
plt.show()
```

```python
# barchart
x = ['A','B','C','D']
y = [15,25,35,45]
plt.bar(x,y)
plt.title("Bar-Chart Example")
plt.show()
```

```python
# data
import random
data = [random.randint(1,30) for i in range(500)]
plt.hist(data, bins=30)
plt.title("Histogram-chart Example")
plt.show()
```

```python
languages = ["Pyhon", "C#", "Java", "JavaScript"]
students = [40, 30,35, 20]
plt.pie(students, labels=languages, startangle=90,autopct="%1.1f%%")
plt.title("Student Fav Programming Language")
plt.show()
```

```python
y1 = [10,15,20,30,35]
y2 = [15,25,40,45,50]
plt.scatter(y1,y2)
plt.title("Scatter plot Example")
plt.show()
```

```python
# data-1 for bar chart
days = ["Mon","Wed","Thr","Fri"]
sales = [20,40,35,55]

# data-2 for scatter plot
y1 = [10,15,20,30,35]
y2 = [15,25,40,45,50]

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.bar(days,sales)
plt.title("Daily Sales")
plt.xlabel("Week")
plt.ylabel("Sales")

plt.subplot(1,2,2)
plt.scatter(y1,y2)
plt.title("sub plot example")
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.show()
```

```python
import pandas as pd
```

```python
data = {
    "month":['Jan','Feb','March','April'],
    "Sales":[50,40,60,67]
}
df = pd.DataFrame(data)
df
```

```python
plt.bar(df["month"], df["Sales"])
plt.title("Monthly-Sales Report")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```
