```python
import sqlite3
conn = sqlite3.connect('test.db')
print("database connection succed....")
```

```python
cursor = conn.cursor()

# create table
cursor.execute("""
    create table if  not exists Student(
    Id integer primary key autoincrement,
    Name text,
    Age integer,
    City text
    )
""")
print("Student table Created Successfully...")
```

```python
cursor.execute("""
    insert into Student(Name,Age,City)values
    (?,?,?)
""",("Ali",23,"Karachi"))
conn.commit()
print("data inserted succesfully...")
```

```python
cursor.execute("""
    select * from Student
""")
data = cursor.fetchall()
print(data)
```

```python
std_data  = [
    ("Ahmad",20,"Karachi"),
    ("Aysha",19,"Lahore"),
    ("Bilal",21,"Swat"),
    ("Waqar",25,"Karachi")
]
cursor.executemany("""
    insert into Student(Name,Age,City)values(?,?,?)
""",(std_data))
conn.commit()
print("data inserted succesfully...")
```

```python
cursor.execute("select * from Student where Age >= 22    ")
data = cursor.fetchall()
for std in data:
    print(std)
```

```python
cursor.execute("""
    update Student
    set City=?, Age=?
    where Id=?
""",("Lahore",20,1))
conn.commit()

cursor.execute("select * from Student")
data = cursor.fetchall()
print(data)
```

```python
# delete Data
cursor.execute("""
    delete from Student
    where Id=?
""",(1,))
conn.commit()
cursor.execute("select * from Student")
data = cursor.fetchall()
print(data)
```
