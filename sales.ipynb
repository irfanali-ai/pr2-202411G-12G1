```python
import numpy as np
```

```python
data = np.genfromtxt(
    "data.csv",
    delimiter = ',',
    names = True,
    dtype = None,
    encoding = 'utf-8'
)
print(data)
```

```python
print(data.dtype.names)
```

```python
print(data["Age"])
```

```python
ages = data["Age"]
prices = data["Unit_Price"]
genders = data["Gender"]
quantity = data["Quantity"]
rating = data["Rating"]
product = data["Product"]
```

```python
print(prices)
print(quantity)
```

```python
# calculate Sales
print("/n---Sales----")
sales = prices * quantity
print(sales)
```

```python
print(f"average age : {np.average(ages)}")
print(f"youngest age: {np.min(ages)}")
print(f"oldest age : {np.max(ages)}")
```

```python
print(f"average rating: {np.mean(rating)}")
print(f"max rating: {np.max(rating)}")
print(f"min rating: {np.min(rating)}")
print(f"std in rating : {np.std(rating)}")
```

```python
print(f"average sales: {np.average(sales)}")
print(f"max sale: {np.max(sales)}")
print(f"min sales: {np.min(sales)}")
print(f"std for sales : {np.std(sales)}")
```

```python
print("sum of Male:", np.sum(genders=="Male"))
print(f"sum of Female : {np.sum(genders=='Female')}")
```

```python
print(f"oldest customer: {np.sum(ages>40)}")
```
